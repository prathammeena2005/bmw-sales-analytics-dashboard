Data Cleaning

```python
# Import Libraries
import pandas as pd
import numpy as np
```

```python
# Load the Dataset
df = pd.read_csv("dataset/data_raw.csv")
```

```python
# Creating a copy of the original DataFrame to work on
df_clean = df.copy()
```

```python
# Removing duplicates
print("Duplicate Rows Before:", df_clean.duplicated().sum())
df_clean.drop_duplicates(inplace=True)
print("Duplicate Rows After:", df_clean.duplicated().sum())
```

```python
# Handling Missing Values
print(df_clean.isnull().sum())
```

```python
# Validating Data Types
df_clean.dtypes
```

```python
# Validating Data
print(df_clean.describe().T)
print("Invalid Years:", df_clean[(df_clean["Year"] < 2010) | (df_clean["Year"] > 2024)].shape[0])
print("Negative Prices:", df_clean[df_clean["Price_USD"] < 0].shape[0])
print("Negative Mileage:", df_clean[df_clean["Mileage_KM"] < 0].shape[0])
print("Invalid Engine Size:", df_clean[df_clean["Engine_Size_L"] <= 0].shape[0])
```

```python
# Checking the cleaned DataFrame
print(df_clean.info())
print(df_clean.isnull().sum())
print(df_clean.duplicated().sum())
```

```python
# Exporting the cleaned DataFrame to a new CSV file
df_clean.to_csv("dataset/data_cleaned.csv", index=False)
```
